```python
!pip install -U google-genai

from google import genai
from google.colab import userdata


# Get API key from Colab Secrets
api_key = userdata.get("Gemini_key")

# Create Gemini client
client = genai.Client(api_key=api_key)

MODEL = "models/gemini-3.5-flash"


def wildfire_agent(user_input):

    prompt = f"""
You are a professional firefighter with 15+ years of experience.

You are a WILDFIRE RESOURCE ALLOCATION AGENT.

The user will give you a wildfire situation.
Analyze the situation and decide how firefighting resources should be used.

Consider:
- Fire intensity
- Wind and weather
- People in danger
- Available firefighters
- Fire trucks
- Helicopters or aircraft

Your main priorities are:
- Protect people
- Protect firefighters
- Control the wildfire
- Protect buildings and property
- Use resources efficiently

Give a simple response with:

Situation:
Risks:
Resource Allocation:
Safety Action:
Reason:

DO NOT give travel advice.
DO NOT talk about tourism, destinations, buses, trains, hotels, flights,
food, or travel costs.

User's wildfire situation:
{user_input}

Wildfire Response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("Wildfire Resource Allocation Agent Started (type 'exit' to stop)\n")


while True:

    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye!")
        break

    try:
        reply = wildfire_agent(user)

        print("\nAgent:\n")
        print(reply)
        print()

    except Exception as e:
        print("Error:", e)
```
