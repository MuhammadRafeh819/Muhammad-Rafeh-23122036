```python
"""A small rule-based railway recommendation agent for Pakistan."""

TRAINS = [
    {
        "name": "Green Line Express",
        "source": "Karachi",
        "destination": "Lahore",
        "class": "Business",
        "fare": 8500,
        "departure": "9:00 AM",
    },
    {
        "name": "Karachi Express",
        "source": "Karachi",
        "destination": "Lahore",
        "class": "Economy",
        "fare": 4500,
        "departure": "5:30 PM",
    },
    {
        "name": "Tezgam",
        "source": "Karachi",
        "destination": "Rawalpindi",
        "class": "First",
        "fare": 9000,
        "departure": "7:00 AM",
    },
    {
        "name": "Khyber Mail",
        "source": "Peshawar",
        "destination": "Karachi",
        "class": "Economy",
        "fare": 5000,
        "departure": "8:15 AM",
    },
    {
        "name": "Awam Express",
        "source": "Peshawar",
        "destination": "Karachi",
        "class": "Business",
        "fare": 7800,
        "departure": "4:45 PM",
    },
    {
        "name": "Allama Iqbal Express",
        "source": "Lahore",
        "destination": "Karachi",
        "class": "Economy",
        "fare": 4700,
        "departure": "11:00 AM",
    },
    {
        "name": "Shalimar Express",
        "source": "Lahore",
        "destination": "Karachi",
        "class": "First",
        "fare": 9200,
        "departure": "6:30 PM",
    },
    {
        "name": "Islamabad Express",
        "source": "Lahore",
        "destination": "Islamabad",
        "class": "Business",
        "fare": 4000,
        "departure": "8:00 AM",
    },
    {
        "name": "Margalla Express",
        "source": "Rawalpindi",
        "destination": "Lahore",
        "class": "Economy",
        "fare": 2500,
        "departure": "2:00 PM",
    },
    {
        "name": "Subak Raftar",
        "source": "Lahore",
        "destination": "Rawalpindi",
        "class": "First",
        "fare": 5500,
        "departure": "6:00 AM",
    },
]


def get_user_input():
    """Ask for and normalize the travel details."""
    source = input("Enter source city: ").strip().title()
    destination = input("Enter destination city: ").strip().title()
    train_class = input("Enter class (Economy/Business/First): ").strip().title()

    while True:
        budget_text = input("Enter your maximum budget (PKR): ").strip()
        try:
            budget = int(budget_text)
            if budget < 0:
                print("Budget cannot be negative. Please try again.")
            else:
                break
        except ValueError:
            print("Please enter the budget as a whole number.")

    return source, destination, train_class, budget


def find_train(source, destination, requested_class, budget):
    """Apply the route, class, and budget rules in order."""
    route_found = False
    class_found = False
    over_budget_train = None

    for train in TRAINS:
        if train["source"] != source or train["destination"] != destination:
            continue
        route_found = True

        if train["class"] != requested_class:
            continue
        class_found = True

        if train["fare"] <= budget:
            reasons = [
                "✓ Source and destination match the train route",
                "✓ Train class matches the requested class",
                "✓ Fare is within your budget",
            ]
            return train, reasons, None

        over_budget_train = train

    if not route_found:
        hint = "Hint: no train covers that route."
    elif not class_found:
        hint = "Hint: the route exists, but not in the requested class."
    else:
        hint = "Hint: matching trains were found, but they exceed your budget."

    return None, [], (over_budget_train, hint)


def show_result(train, reasons, failure):
    """Display either the first recommendation or a useful failure message."""
    if train is not None:
        print("\nRecommended train")
        print(f"Name: {train['name']}")
        print(f"Departure: {train['departure']}")
        print(f"Class: {train['class']}")
        print(f"Fare: PKR {train['fare']}")
        print("Why this train matches:")
        for reason in reasons:
            print(reason)
        return

    over_budget_train, hint = failure
    print("\nNo suitable train was found")
    if over_budget_train is not None:
        print(
            f"{over_budget_train['name']} matches the route and class, "
            f"but its fare of PKR {over_budget_train['fare']} exceeds your budget."
        )
    print(hint)


def main():
    print("Pakistan Railway Rule-Based Agent")
    source, destination, requested_class, budget = get_user_input()
    train, reasons, failure = find_train(
        source, destination, requested_class, budget
    )
    show_result(train, reasons, failure)


main()


# Demo scenarios:
# 1. Karachi -> Lahore, Economy, 5000: recommends Karachi Express.
# 2. Lahore -> Karachi, First, 5000: matching train exceeds the budget.
# 3. Islamabad -> Karachi, Economy, 10000: no train covers that route.
```
